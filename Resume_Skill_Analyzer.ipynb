```python
import PyPDF2

def extract_text(pdf_file):
    reader = PyPDF2.PdfReader(pdf_file)
    text = ""

    for page in reader.pages:
        text += page.extract_text()

    return text
```

```python
skills = [
    "python", "excel", "sql", "machine learning",
    "data analysis", "communication", "teamwork",
    "html", "css", "javascript", "power bi"
]
```

```python
def analyze_resume(text, skills):
    text = text.lower()
    found_skills = []

    for skill in skills:
        if skill in text:
            found_skills.append(skill)

    return found_skills
```

```python
file_name = list(uploaded.keys())[0]

text = extract_text(file_name)
found_skills = analyze_resume(text, skills)

print("FOUND SKILLS:")
print(found_skills)

print("\nMISSING SKILLS:")
print(set(skills) - set(found_skills))
```

```python
def calculate_score(found_skills, skills):
    score = (len(found_skills) / len(skills)) * 100
    return round(score, 2)

score = calculate_score(found_skills, skills)

print("\n===== RESUME SCORE =====")
print("Score:", score, "/100")
```
